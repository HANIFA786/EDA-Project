```python
import pandas as pd
d = pd.read_csv("/content/adult.csv")
d["Hours per Week"]=d["Hours per Week"].bfill()
print(d["Hours per Week"])
```

```python
Q1=d["Hours per Week"].quantile(0.25)
Q3=d['Hours per Week'].quantile(0.75)
IQR=Q3-Q1
lower_bound=(Q1-1.5*IQR)
upper_bound=Q3+1.5*IQR
print(lower_bound)
print(upper_bound)
```

```python
from scipy.stats import zscore
d["Age_zscore"]=zscore(d["Age"])
print(d["Age_zscore"])
z_outliers=d[d["Age_zscore"].abs()>3]
print("Z-Score Outliers:",len(z_outliers))
```

```python
from scipy.stats.mstats import winsorize
d["Age_winsorized"]=winsorize(d["Age"],limits=[0,0.5])
print(d.head())
```

```python
import pandas as pd
data=pd.read_csv("/content/adult.csv")
print(data)
```

```python
import matplotlib.pyplot as plt
data['Age'].hist(bins=20, edgecolor='black', figsize=(8,5))
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt

gender_counts = data['Gender'].str.strip().value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%'
)
plt.title('Gender Distribution')
plt.show()
```
